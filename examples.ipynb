# RAGMechanic: End-to-End RAG Example

This notebook demonstrates how to use RAGMechanic to build a Retrieval-Augmented Generation (RAG) pipeline, query documents, evaluate the pipeline, and generate an HTML report.

## Workflow

1. Load documents.
2. Configure embeddings, an LLM, and a vector store.
3. Ask a question against the documents.
4. Evaluate the pipeline.
5. Generate an HTML report.

## Prerequisites

Install RAGMechanic and the integrations required by this example. Configure the relevant API key in your environment before running the notebook.


```python
from pathlib import Path
from dotenv import load_dotenv
from ragmechanic import RAGPipeline

load_dotenv()

ROOT = Path.cwd()
docs = [
    "../ragmechanic/documents/Machine Learning.pdf",
    "../ragmechanic/documents/AI.txt",
]
```

```python
rag = RAGPipeline(
    docs=docs,
    embedding_model="BAAI/bge-small-en-v1.5",
    llm_provider="google",
    llm_model="gemini-3.5-flash-lite",
    vector_store="chroma",
    chunk_size=1000,
    chunk_overlap=200,
    k=6,
)
```

```python
result = rag.ask("what is machine learning")
```

```python
print(result)
```

```python
result.answer
```

```python
rag.evaluate()
```

```python
rag.save_html()
```
